La funzione **levenshtein_distance** calcola la distanza di Levenshtein tra due stringhe, ovvero il numero minimo di operazioni elementari necessarie per trasformare una stringa iniziale s in una stringa di destinazione t. Le operazioni considerate sono: inserimento, cancellazione e sostituzione di caratteri.

```python
def levenshtein_distance(s, t):

    """
    Funzione "levenshtein_distance"

    - parametro s: stringa iniziale
    - param t: stringa di destinazione
    - return: distanza di Levenshtein tra s e t
    """

    m, n = len(s), len(t)

    # Creazione della matrice (m+1)x(n+1) inizializzata con zeri
    dp = []
    for _ in range(m+1):
      row = []
      for _ in range(n+1):
        row.append(0)
      dp.append(row)

    # Inizializzazione della prima riga e della prima colonna
    # La prima colonna (dp[i][0]) tiene traccia del costo di cancellazione di i caratteri dalla stringa iniziale per ottenere una stringa vuota
    for i in range(m+1):
        dp[i][0] = i
    # La prima riga (dp[0][j]) tiene traccia del costo di inserimento di j caratteri per trasformare una stringa vuota nella sottostringa di destinazione di lunghezza j
    for j in range(n+1):
        dp[0][j] = j

    # Popolamento della matrice
    for i in range(1, m+1):
        for j in range(1, n+1):
            if s[i-1] == t[j-1]: # Se i caratteri di s e t sono uguali, nessun costo aggiuntivo
              dp[i][j] = dp[i-1][j-1]
            else: # Se i caratteri di s e t sono diversi, allora calcola il costo minimo tra cancellazione, inserimento e sostituzione
               dp[i][j] = 1 + min(dp[i-1][j], # Cancellazione
                               dp[i][j-1],    # Inserimento
                               dp[i-1][j-1])  # Sostituzione

    # La cella in basso a destra contiene la distanza di Levenshtein tra s e t
    return dp[m][n]
```

La funzione **suggest_correction** ha lo scopo di verificare se una determinata parola è presente nel dizionario fornito. Se la parola è già presente, viene restituita direttamente. In caso contrario, la funzione confronta la query con tutte le parole presenti nel dizionario calcolando, per ciascuna di esse, la distanza di Levenshtein. La parola con la distanza minima viene selezionata come correzione suggerita e restituita come output.

```python
import sys

def suggest_correction(query, dictionary):

    """
    Funzione "suggestion_correction"

    - parametro query: parola da controllare
    - parametro dictionary: lista di parole da confrontare con la parola da controllare
    - return: parola corretta
    """

    # Normalizzazione in minuscolo, per il confronto, sia della query inserita dall'utente che del dizionario di parole
    query_lower = query.lower()

    dictionary_lower = []
    for word in dictionary:
      dictionary_lower.append(word.lower())

    # Se la parola inserita dall'utente è presente nel dizionario, allora verrà restituita
    if query_lower in dictionary_lower:
        return dictionary[dictionary_lower.index(query_lower)]

    # Se la parola inserita dall'utente non è presente nel dizionario, allora si calcola la distanza di Levenshtein per ogni parola del dizionario
    best_word = None
    best_distance = sys.maxsize # Inizializzazione della miglior distanza con un valore molto grande
    for idx, word in enumerate(dictionary_lower): # Per ogni parola presente nel dizionario, calcolare la distanza rispetto alla query dell'utente
        distance = levenshtein_distance(query_lower, word)
        if distance < best_distance: # Se questa nuova distanza è più piccola di quella trovata finora, la si memorizza insieme alla nuova parola
            best_distance = distance
            best_word = dictionary[idx] # Memorizzazione della parola originale dalla lista 'dictionary'

    # Ritorno della parola con la distanza minore trovata, suggerendo la correzione
    return best_word
```

* Parole con trasposizioni di lettere, ovvero caratteri adiacenti scambiati di posizione;
* Parole con lettere mancanti o inserite in eccesso.
